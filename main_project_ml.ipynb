```python
x = data.drop('hazardous', axis=1)
y = data['hazardous']
```

```python
#smote
```

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE()

x_smote, y_smote = sm.fit_resample(x, y)
```

```python
y_smote.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

col= x_smote.columns

data_scale = scaler.fit_transform(x_smote)

x_smote = pd.DataFrame(data_scale, columns=col)
```

```python
#train test
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_smote, y_smote, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

lr = LogisticRegression()

lr.fit(x_train , y_train )

y_pred = lr.predict(x_test)


print(classification_report(y_pred, y_test))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report

dt = DecisionTreeClassifier()

dt.fit(x_train , y_train )

y_pred = dt.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

rf =  RandomForestClassifier()

rf.fit(x_train , y_train )

y_pred = rf.predict(x_test)

print(confusion_matrix(y_pred, y_test))
print(classification_report(y_pred, y_test))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report

knn = KNeighborsClassifier()

knn.fit(x_train , y_train )

y_pred = knn.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report
nb=GaussianNB()
nb.fit(x_train , y_train )

y_pred = nb.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report
gb=GradientBoostingClassifier()
gb.fit(x_train , y_train )

y_pred = gb.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import classification_report
ab=AdaBoostClassifier()
ab.fit(x_train , y_train )

y_pred = ab.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report
xgb=XGBClassifier()
xgb.fit(x_train , y_train )

y_pred = xgb.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report
svc=SVC()
svc.fit(x_train , y_train )

y_pred = svc.predict(x_test)

print(classification_report(y_pred, y_test))
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

# Page settings
st.set_page_config(
    page_title="NEO Hazard Prediction",
    page_icon="🌍",
    layout="centered"
)

# Load model and scaler
@st.cache_resource
def load_files():
    model = joblib.load("random_forest_model.pkl")
    scaler = joblib.load("standard_scaler.pkl")
    feature_columns = joblib.load("feature_columns.pkl")

    return model, scaler, feature_columns


model, scaler, feature_columns = load_files()

# Application title
st.title("🌍 Near-Earth Object Hazard Prediction")

st.write(
    "Enter the details of a Near-Earth Object "
    "to predict whether it is hazardous or non-hazardous."
)

st.divider()

# Input section
st.subheader("Enter NEO Details")

est_diameter_min = st.number_input(
    "Estimated Diameter Minimum",
    min_value=0.0,
    value=0.1,
    step=0.01
)

est_diameter_max = st.number_input(
    "Estimated Diameter Maximum",
    min_value=0.0,
    value=0.2,
    step=0.01
)

relative_velocity = st.number_input(
    "Relative Velocity",
    min_value=0.0,
    value=50000.0,
    step=100.0
)

miss_distance = st.number_input(
    "Miss Distance",
    min_value=0.0,
    value=50000000.0,
    step=100000.0
)

absolute_magnitude = st.number_input(
    "Absolute Magnitude",
    value=20.0,
    step=0.1
)

st.divider()

# Prediction button
if st.button("Predict Hazard Status", use_container_width=True):

    # Create input DataFrame
    input_data = pd.DataFrame(
        [[
            est_diameter_min,
            est_diameter_max,
            relative_velocity,
            miss_distance,
            absolute_magnitude
        ]],
        columns=[
            "est_diameter_min",
            "est_diameter_max",
            "relative_velocity",
            "miss_distance",
            "absolute_magnitude"
        ]
    )

    # Arrange columns in the same order as training
    input_data = input_data[feature_columns]

    # Apply the fitted scaler
    input_data_scaled = scaler.transform(input_data)

    # Make prediction
    prediction = model.predict(input_data_scaled)

    predicted_value = prediction[0]

    st.subheader("Prediction Result")

    if predicted_value == 1:
        st.error(
            "⚠️ This Near-Earth Object is predicted to be HAZARDOUS."
        )
    else:
        st.success(
            "✅ This Near-Earth Object is predicted to be NON-HAZARDOUS."
        )

    # Prediction probability
    if hasattr(model, "predict_proba"):

        probabilities = model.predict_proba(input_data_scaled)

        probability_df = pd.DataFrame(
            {
                "Class": [
                    "Non-Hazardous",
                    "Hazardous"
                ],
                "Probability": [
                    probabilities[0][0] * 100,
                    probabilities[0][1] * 100
                ]
            }
        )

        st.subheader("Prediction Probability")

        st.bar_chart(
            probability_df.set_index("Class")
        )
```
