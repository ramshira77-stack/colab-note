```python
from sklearn.model_selection import cross_val_score
```

```python
models = {
    "Logistic Regression": LogisticRegression(),
    "Decisiontree" : DecisionTreeClassifier(random_state=42),
    "Random Forest" : RandomForestClassifier(random_state=42),
    "knn":KNeighborsClassifier(n_neighbors=7),
    "svm":SVC(kernel='rbf',C=1.0,gamma='scale',random_state=42),
    "Naive Bayes":GaussianNB(),
    "XGBoost":XGBClassifier(n_estimators=200,max_depth=7,learning_rate=0.1,random_state=42),
    "AdaBoost":AdaBoostClassifier(n_estimators=100,estimator=tree,learning_rate=1.0,random_state=42),
    "GradientBoost":GradientBoostingClassifier(n_estimators=100,learning_rate=0.1,random_state=42)
}

for name, model in models.items():
  if name in ['Logistic Regression','knn']:
    scores =  cross_val_score(model,x_smote,y_smote,cv=5)
  else:
    scores = cross_val_score(model,x,y,cv=5)

  print(f'{name}:')
  print("cv score =",scores)
  print("mean accuracy=",scores.mean())
  print("-"*40)
```
